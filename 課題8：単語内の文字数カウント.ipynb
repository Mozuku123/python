```python
words=[]
dictionary={'a':0, 'b':0, 'c':0, 'd':0, 'e':0, 'f':0, 'g':0, 'h':0, 'i':0, 'j':0, 'k':0, 'l':0, 'm':0, 'n':0, 'o':0, 'p':0, 'q':0, 'r':0, 's':0, 't':0, 'u':0, 'v':0, 'w':0, 'x':0, 'y':0, 'z':0}

while True:
    word = input("英単語を入力してください：")
    if word == "":
        break
    words.append(word)
    for char in word.lower(): 
        if char in dictionary:
            dictionary[char] += 1
print(f"入力した英単語：{sorted(words)}")

# 1文字以上あったすべての文字について表示
for letter in sorted(dictionary.keys()):
    if dictionary[letter] > 0:
        print(f"{letter}が{dictionary[letter]}個ありました")
```
