```python
class Baseball_Team:
    def __init__(self, name, win, lose,draw):
        self.name = name
        self.win= win
        self.lose=lose
        self.draw=draw
    
    def calc_win_rate(self):
        return self.win/(self.win+self.lose)
    def show_team_result(self):
        return f"{self.name},{self.win},{self.lose},{self.draw},{self.calc_win_rate()}"

archers=Baseball_Team("Archers",77,64,2)
meteors=Baseball_Team("Meteors",71,69,3)
cheetahs=Baseball_Team("Cheetahs",69,68,6)
sharks=Baseball_Team("Sharks",70,70,3)
pythons=Baseball_Team("Pythons",68,73,2)
owls=Baseball_Team("Owls",59,82,2)

print("name","win","lose","draw","rate")
print(archers.show_team_result())
print(meteors.show_team_result())
print(cheetahs.show_team_result())
print(sharks.show_team_result())
print(pythons.show_team_result())
print(owls.show_team_result())
```
